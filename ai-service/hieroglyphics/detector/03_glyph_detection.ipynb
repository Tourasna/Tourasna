```python
# ── Cell 11: Convert COCO → YOLO format ───────────────
import json, os, shutil
from PIL import Image

YOLO_DIR = '/content/hieroglyph_yolo'
os.makedirs(f'{YOLO_DIR}/images/train', exist_ok=True)
os.makedirs(f'{YOLO_DIR}/images/val', exist_ok=True)
os.makedirs(f'{YOLO_DIR}/labels/train', exist_ok=True)
os.makedirs(f'{YOLO_DIR}/labels/val', exist_ok=True)

def coco_to_yolo(coco_data, image_dir, split, class_map):
    """Convert COCO annotations to YOLO format"""
    # Build image lookup
    img_lookup = {img['id']: img for img in coco_data['images']}

    # Group annotations by image
    from collections import defaultdict
    ann_by_img = defaultdict(list)
    for ann in coco_data['annotations']:
        ann_by_img[ann['image_id']].append(ann)

    converted = 0
    skipped = 0

    for img_id, img_info in img_lookup.items():
        fname = img_info['file_name']
        src_path = os.path.join(image_dir, fname)

        if not os.path.exists(src_path):
            skipped += 1
            continue

        w, h = img_info['width'], img_info['height']
        if w == 0 or h == 0:
            skipped += 1
            continue

        # Copy image
        dst_name = f"{split}_{img_id}_{os.path.basename(fname)}"
        dst_name = dst_name.replace(' ', '_')  # handle spaces
        shutil.copy(src_path, f'{YOLO_DIR}/images/{split}/{dst_name}')

        # Write YOLO labels
        label_name = os.path.splitext(dst_name)[0] + '.txt'
        with open(f'{YOLO_DIR}/labels/{split}/{label_name}', 'w') as f:
            for ann in ann_by_img.get(img_id, []):
                cat_name = None
                for cat in coco_data['categories']:
                    if cat['id'] == ann['category_id']:
                        cat_name = cat['name']
                        break

                if cat_name not in class_map:
                    continue

                cls_id = class_map[cat_name]
                bx, by, bw, bh = ann['bbox']  # COCO: x,y,w,h (top-left)

                # Convert to YOLO: cx, cy, w, h (normalized)
                cx = (bx + bw / 2) / w
                cy = (by + bh / 2) / h
                nw = bw / w
                nh = bh / h

                # Clamp to [0, 1]
                cx = max(0, min(1, cx))
                cy = max(0, min(1, cy))
                nw = max(0, min(1, nw))
                nh = max(0, min(1, nh))

                f.write(f"{cls_id} {cx:.6f} {cy:.6f} {nw:.6f} {nh:.6f}\n")

        converted += 1

    return converted, skipped

# ── Convert segmentation dataset (glyphs) ─────────────
# Map all sign types to class 0 (glyph)
seg_class_map = {'Right_Sign': 0, 'Left_Sign': 0, 'Dual_Sign': 0}

seg_train_conv, seg_train_skip = coco_to_yolo(
    train_ann, '/content/segmentation', 'train', seg_class_map
)
seg_val_conv, seg_val_skip = coco_to_yolo(
    val_ann, '/content/segmentation', 'val', seg_class_map
)

print("=== Segmentation (Glyphs) ===")
print(f"  Train: {seg_train_conv} converted, {seg_train_skip} skipped")
print(f"  Val:   {seg_val_conv} converted, {seg_val_skip} skipped")

# ── Convert layout dataset (lines) ────────────────────
# Only take 'Line' as class 1
layout_class_map = {'Line': 1, 'Cartoush': 1}

lay_train_conv, lay_train_skip = coco_to_yolo(
    layout_train, '/content/layout', 'train', layout_class_map
)
lay_val_conv, lay_val_skip = coco_to_yolo(
    layout_val, '/content/layout', 'val', layout_class_map
)

print(f"\n=== Layout (Lines) ===")
print(f"  Train: {lay_train_conv} converted, {lay_train_skip} skipped")
print(f"  Val:   {lay_val_conv} converted, {lay_val_skip} skipped")

# ── Summary ───────────────────────────────────────────
train_imgs = len(os.listdir(f'{YOLO_DIR}/images/train'))
val_imgs = len(os.listdir(f'{YOLO_DIR}/images/val'))
print(f"\n=== YOLO Dataset Ready ===")
print(f"  Train images: {train_imgs}")
print(f"  Val images:   {val_imgs}")

# ── Write dataset.yaml ────────────────────────────────
yaml_content = f"""path: {YOLO_DIR}
train: images/train
val: images/val
nc: 2
names: ['glyph', 'line']
"""

with open(f'{YOLO_DIR}/dataset.yaml', 'w') as f:
    f.write(yaml_content)

print(f"\n  dataset.yaml written ✅")
```
